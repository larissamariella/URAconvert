```python
import os
import shutil

pasta = "audios_sac"

if os.path.exists(pasta):
    for arquivo in os.listdir(pasta):
        caminho = os.path.join(pasta, arquivo)
        if os.path.isfile(caminho):
            os.remove(caminho)
    print("Pasta 'audios_ulaw' limpa com sucesso.")
else:
    print("A pasta 'audios_ulaw' não existe.")
```

```python
import shutil
import os

pasta = "audios_ulaw"

if os.path.exists(pasta):
    shutil.rmtree(pasta)

os.makedirs(pasta, exist_ok=True)

print("Pasta 'audios_ulaw' recriada do zero.")
```

```python
import os

extensoes = (".mp3", ".wav", ".ogg", ".flac", ".aac", ".m4a")

for arquivo in os.listdir("."):
    if arquivo.lower().endswith(extensoes):
        os.remove(arquivo)

print("Arquivos de entrada removidos.")
```

```python
import os
import subprocess

# Pasta de saída
output_dir = "audios_sac_novos"
os.makedirs(output_dir, exist_ok=True)

# Extensões aceitas
extensoes = (".mp3", ".wav", ".ogg", ".flac", ".aac", ".m4a")

for arquivo in os.listdir("."):
    if arquivo.lower().endswith(extensoes):
        nome_base = os.path.splitext(arquivo)[0]
        saida = os.path.join(output_dir, f"{nome_base}.wav")

        comando = [
            "ffmpeg",
            "-y",
            "-i", arquivo,
            "-ac", "1",                 # mono
            "-ar", "8000",              # 8 kHz
            "-acodec", "pcm_mulaw",     # CCITT u-Law
            saida
        ]

        subprocess.run(comando, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        print(f"Convertido: {arquivo} → {saida}")
```
